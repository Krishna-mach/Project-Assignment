```python
#Wine Dataset Assignment Solutions
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine

# Load dataset
data = load_wine(as_frame=True)
df = data.frame
```

```python
#Q1.Line Graph (Alcohol vs Malic Acid)
plt.plot(df['alcohol'], df['malic_acid'], marker='o')
plt.xlabel('Alcohol')
plt.ylabel('Malic Acid')
plt.title('Alcohol vs Malic Acid')
plt.show()
```
