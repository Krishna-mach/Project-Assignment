```python
#Dataset Used: Wine dataset from sklearn.datasets
#“How to Load the Dataset
#data = load_wine(as_frame=True)
#df = data.frame”
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_wine

# Load dataset
data = load_wine(as_frame=True)
df = data.frame
```

```python
#Q1. Plot a line graph showing the relationship between alcohol and malic_acid using
#matplotlib.pyplot.
plt.plot(df['alcohol'], df['malic_acid'], marker='o')
plt.xlabel('Alcohol')
plt.ylabel('Malic Acid')
plt.title('Alcohol vs Malic Acid')
plt.show()
```

```python
#Q2. Create a boxplot of flavanoids grouped by wine class (target) using
#seaborn.boxplot().
sns.boxplot(x='target', y='flavanoids', data=df)
plt.title('Flavanoids grouped by Wine Class')
plt.show()
```

```python
#Q3. Draw a 2D scatterplot for alcohol vs color_intensity, with color coding based on
#wine class.
sns.scatterplot(x='alcohol', y='color_intensity', hue='target', data=df, palette='Set1')
plt.title('Alcohol vs Color Intensity by Wine Class')
plt.show()
```

```python
#Q4. Create a 3D scatterplot using alcohol, color_intensity, and hue as axes. Use
#different colors for each wine class. Use matplotlib's 3D plotting tools.
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

for wine_class in df['target'].unique():
    subset = df[df['target'] == wine_class]
    ax.scatter(subset['alcohol'], subset['color_intensity'], subset['hue'], label=f'Class {wine_class}')

ax.set_xlabel('Alcohol')
ax.set_ylabel('Color Intensity')
ax.set_zlabel('Hue')
ax.legend()
plt.title('3D Scatterplot of Wine Features')
plt.show()
```

```python
#Q5. Select any 4 numerical columns from the dataset and create a seaborn.pairplot()
#with hue='target'.
sns.pairplot(df[['alcohol','malic_acid','flavanoids','color_intensity','target']], hue='target')
plt.show()
```

```python
#Q7. Plot a histogram of the alcohol feature. Overlay the PDF (Probability Density
#Function) using Seaborn’s kdeplot.
sns.histplot(df['alcohol'], kde=False, bins=15, color='skyblue')
sns.kdeplot(df['alcohol'], color='red')
plt.title('Histogram + PDF of Alcohol')
plt.show()
```

```python
#Q8. Perform a univariate analysis on the malic_acid feature using its PDF plot.
#Comment on its distribution shape.
sns.kdeplot(df['malic_acid'], shade=True, color='green')
plt.title('PDF of Malic Acid')
plt.show()
```

```python
#Q9. Compute and plot the CDF of the proline feature. What percentage of samples
#have proline less than 800?
# Compute CDF
proline_sorted = np.sort(df['proline'])
cdf = np.arange(1, len(proline_sorted)+1) / len(proline_sorted)

plt.plot(proline_sorted, cdf, marker='.')
plt.xlabel('Proline')
plt.ylabel('CDF')
plt.title('CDF of Proline')
plt.show()

# Percentage of samples with proline < 800
percentage = (df['proline'] < 800).mean() * 100
print(f"Percentage of samples with proline < 800: {percentage:.2f}%")
```
