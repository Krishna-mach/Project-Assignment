```python
#Ques.1. Write a Pandas program to create a DataFrame from a dictionary of lists, 
#where: 
#• One column contains random integers between 10 and 100 (size=10). 
#• Another column contains random float values between 0 and 1. 
#• Extract the rows where the integer column values are divisible by 5 and the 
#float column values are greater than 0.5. 

import pandas as pd
import numpy as np

# Create DataFrame
df = pd.DataFrame({
    'Integers': np.random.randint(10, 101, size=10),
    'Floats': np.random.rand(10)
})

# Filter rows
result = df[(df['Integers'] % 5 == 0) & (df['Floats'] > 0.5)]
print(result)

```

```python
#Ques.2. Given a DataFrame with missing values: 
#data = {'A': [1, 2, np.nan, 4, 5], 
#'B': [5, np.nan, np.nan, 8, 10], 
#'C': ['a', 'b', np.nan, 'd', 'e']} 
#df = pd.DataFrame(data) 
#•  Fill missing values in column A with the mean of the column. 
#•  Replace missing values in column B with the median. 
#•  Drop rows where column C has missing values. 

import pandas as pd
import numpy as np

data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [5, np.nan, np.nan, 8, 10],
    'C': ['a', 'b', np.nan, 'd', 'e']
}
df = pd.DataFrame(data)

# Fill A with mean
df['A'].fillna(df['A'].mean(), inplace=True)

# Fill B with median
df['B'].fillna(df['B'].median(), inplace=True)

# Drop rows where C is missing
df.dropna(subset=['C'], inplace=True)

print(df)
```

```python
#Ques.3. Given a DataFrame of sales data: 
#data = {'Region': ['North', 'South', 'North', 'South', 'East'], 
#'Product': ['A', 'A', 'B', 'B', 'C'], 
#'Sales': [200, 150, 400, 300, 100]} 
#df = pd.DataFrame(data) 
#•  Pivot the table to show total sales for each product across regions. 
#product. 
data = {
    'Region': ['North', 'South', 'North', 'South', 'East'],
    'Product': ['A', 'A', 'B', 'B', 'C'],
    'Sales': [200, 150, 400, 300, 100]
}
df = pd.DataFrame(data)

# Pivot table
pivot = df.pivot_table(values='Sales', index='Product', columns='Region', aggfunc='sum', fill_value=0)

# Add percentage column
pivot['% of Total'] = (pivot.sum(axis=1) / pivot.sum().sum()) * 100

print(pivot)
```

```python
#Ques.4.  Write a custom function that: 
#• Takes a numeric value as input and returns 'High' if the value is above the mean of 
#the column, otherwise 'Low'. 
#column. 
import pandas as pd
import numpy as np

# Random integers
df = pd.DataFrame({'Values': np.random.randint(1, 100, size=10)})

# Custom function
def classify(x, mean_val):
    return 'High' if x > mean_val else 'Low'

mean_val = df['Values'].mean()
df['Category'] = df['Values'].apply(lambda x: classify(x, mean_val))

print(df)

```

```python
#Ques.5. Create a Pandas Series of random integers between 1 and 100. 
#Perform the following tasks: 
#original series are replaced with the string 'High' and the rest are replaced with 
#2. Compute and display the cumulative sum of the series values. 
import pandas as pd
import numpy as np

# Random series
s = pd.Series(np.random.randint(1, 101, size=10))

# Replace values based on 75th percentile
threshold = s.quantile(0.75)
new_series = s.apply(lambda x: 'High' if x > threshold else 'Low')

# Cumulative sum
cumsum = s.cumsum()

print("Original Series:\n", s)
print("\nHigh/Low Series:\n", new_series)
print("\nCumulative Sum:\n", cumsum)

```
