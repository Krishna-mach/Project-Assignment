```python
#1. Write a program to find the second largest number in a list without using built-in 
#functions like max(). 
nums = [10, 20, 4, 45, 99]
largest = second = float('-inf')

for n in nums:
    if n > largest:
        second = largest
        largest = n
    elif n > second and n != largest:
        second = n

print("Second largest:", second)


```

```python
#2. How would you remove duplicates from a list while maintaining the original order? 
nums = [1, 2, 2, 3, 4, 3]
seen, result = set(), []
for n in nums:
    if n not in seen:
        seen.add(n)
        result.append(n)
print(result)

```

```python
#3. Write a program to reverse a list in-place without using reverse() or slicing. 
nums = [1, 2, 3, 4]
i, j = 0, len(nums)-1
while i < j:
    nums[i], nums[j] = nums[j], nums[i]
    i += 1; j -= 1
print(nums)

```

```python
#4. Given a list of integers, write code to split it into even and odd numbers.
nums = [1, 2, 3, 4, 5, 6]
evens, odds = [], []
for n in nums:
    (evens if n % 2 == 0 else odds).append(n)
print("Evens:", evens, "Odds:", odds)


```

```python
#10. How would you count vowels and consonants in a string? 
s = "Hello World"
vowels = "aeiouAEIOU"
v, c = 0, 0
for ch in s:
    if ch.isalpha():
        if ch in vowels: v += 1
        else: c += 1
print("Vowels:", v, "Consonants:", c)

```

```python
#17. Write a function to compress a string by replacing repeating characters with count: 
#Example: "aaabbc" → "a3b2c1" 
def compress(s):
    result = ""
    i = 0
    while i < len(s):
        count = 1
        while i+1 < len(s) and s[i] == s[i+1]:
            count += 1; i += 1
        result += s[i] + str(count)
        i += 1
    return result

print(compress("aaabbc"))  # a3b2c1
```

```python
#37. Create two sets: one from vowels in a string, and another from consonants — then 
#print both.
s = "Hello World"
vowels = set("aeiouAEIOU")
v_set = {ch for ch in s if ch in vowels}
c_set = {ch for ch in s if ch.isalpha() and ch not in vowels}
print("Vowels:", v_set, "Consonants:", c_set)
```
