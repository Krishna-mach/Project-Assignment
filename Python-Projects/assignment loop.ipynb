```python
#Code on loop

#Ques.1. Print numbers from 1 to 10. If a number is even, break the loop using a for loop 

for i in range(1, 11):
    if i % 2 == 0:
        print(i, "is even → breaking loop")
        break
    else:
        print(i)
else:
    print("Loop finished without break")

```

```python
#Ques.2. Print numbers from 1 to 5, except 3 using a for loop and continue statement. 

for i in range(1, 6):
    if i == 3:
        continue
    print(i)

```

```python
#Ques.3. Print numbers in a list until a negative number is encountered using 
#a while loop.

numbers = [5, 7, 3, -2, 9]
i = 0
while i < len(numbers) and numbers[i] >= 0:
    print(numbers[i])
    i += 1

```

```python
#Ques.4. Find the common elements in two lists using a for loop. 
#list1 = [1,2,3,4,5] 
#list2 = [3,4,5,6,7]

list1 = [1,2,3,4,5]
list2 = [3,4,5,6,7]

for x in list1:
    if x in list2:
        print(x)


```

```python
#Ques.5. Print the Fibonacci sequence up to the 10th term using a while loop. 

a, b = 0, 1
count = 0
while count < 10:
    print(a)
    a, b = b, a + b
    count += 1
```

```python
#Ques.6. Print numbers divisible by 3 or 5 from 1 to 20 using a for loop. 

for i in range(1, 21):
    if i % 3 == 0 or i % 5 == 0:
        print(i)
```

```python
#Ques.7. Find all prime numbers between 1 and 50 using nested for and if.

for num in range(2, 51):
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
```

```python
#Ques.8. Find the first occurrence of a number in a list using a while loop.
numbers = [4, 7, 2, 9, 2, 5]
target = 2
i = 0
while i < len(numbers):
    if numbers[i] == target:
        print("First occurrence at index:", i)
        break
    i += 1
```

```python
#Ques.9. Calculate factorial of a number using a while loop. 

n = 5
fact = 1
i = 1
while i <= n:
    fact *= i
    i += 1
print("Factorial:", fact)
```

```python
#Ques.10. Print below patterns using nested for loops.

for i in range(1, 6):
    print("* " * i)
```

```python
# Upper half
for i in range(1, 6):
    print("* " * i)
# Lower half
for i in range(5, 0, -1):
    print("* " * i)
```
