```python
# Part 1: Arithmetic Operators
#question1
books = 5
price = 150
total_cost = books*price*1.18
print(" Q1 Total cost of books is",total_cost)

#question2
score1, score2, score3 = 85, 92, 78
average = (score1 + score2 + score3) / 3
print("Q2 Average Score:", average)


#question3
print("Q3 Remainder:", 157 % 12)

#question4
print("Q4 Power:", 2 ** 8)

#question5
hours = 7.5
minutes = hours * 60
print("Q5 Total Minutes:", minutes)




```

```python
# Part 2: Logical Operators
#question6
num = 25
print("Q6 Between 10 and 50:", 10 <= num <= 50)

#question7
age, has_permission = 17, True
can_vote = (age >= 18) or has_permission
print("Q7 Voting Eligibility:", can_vote)

#question8
password = "mypass 123"
is_strong = (password != "") and (len(password) > 8)
print("Q8 Strong Password:", is_strong)

#question9
is_weekend, is_holiday = False, True
day_off = is_weekend or is_holiday
print("Q9 Day Off:", day_off)

```

```python
# Part 3: Membership Operators
#question10
word = "apple"
print("Q10 'a' in word:", 'a' in word)

#question11
numbers = [1, 3, 5, 9, 11]
print("Q11 7 not in list:", 7 not in numbers)

#question12
email = "student@yahoo.com"
print("Q12 'gmail' in email:", "gmail" in email)

#question13
word = "rhythm"
vowels = "aeiou"
print("Q13 Any vowel in 'rhythm':", any(v in word for v in vowels))
```

```python
# Part 4: Equality Operators
#question14
marks1, marks2 = 85, 85
print("Q14 Marks Equal:", marks1 == marks2)

#question15
price = 450
print("Q15 Price != 500:", price != 500)

#question16
username, correct_user = "user123", "admin"
print("Q16 Username matches admin:", username == correct_user)

```

```python
# Part 5: Comparison Operators

#question17
temp, limit = 35, 30
print("Q17 Temp > Limit:", temp > limit)

#question18
score1, score2 = 78, 82
print("Q18 Higher Score:", max(score1, score2))

#question19
attendance, minimum = 75, 70
print("Q19 Attendance meets minimum:", attendance >= minimum)

#question20
marks, passing = 65, 70
print("Q20 Marks below passing:", marks < passing)

```

```python
# Part 6: Control Flow - If Statements
#question21
number = -5
if number > 0:
    print("Q21 Positive")
elif number < 0:
    print("Q21 Negative")
else:
    print("Q21 Zero")

#question22
marks = 76
if marks >= 90:
    grade = "A"
elif marks >= 80:
    grade = "B"
elif marks >= 70:
    grade = "C"
elif marks >= 60:
    grade = "D"
else:
    grade = "F"
print("Q22 Grade:", grade)

#question23
light = "Red"
if light == "Red":
    print("Q23 Stop")
elif light == "Yellow":
    print("Q23 Wait")
elif light == "Green":
    print("Q23 Go")
else:
    print("Q23 Invalid")

#question24
age = 25
if age < 12:
    ticket_price = 100
elif age <= 60:
    ticket_price = 200
else:
    ticket_price = 150
print("Q24 Ticket Price:", ticket_price)

#question25
units = 150
if units <= 100:
    bill = units * 2
elif units <= 200:
    bill = (100 * 2) + (units - 100) * 3
else:
    bill = (100 * 2) + (100 * 3) + (units - 200) * 5
print("Q25 Electricity Bill:", bill)

#question26
year = 2024
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print("Q26 Leap Year")
else:
    print("Q26 Not Leap Year")

#question27
username, password = "admin", "pass 123"
correct_user, correct_pass = "admin", "pass 123"
if username == correct_user and password == correct_pass:
    print("Q27 Login Successful")
else:
    print("Q27 Login Failed")

#question28
balance, withdraw_amount = 5000, 6000
if withdraw_amount <= balance:
    print("Q28 Withdrawal Successful")
else:
    print("Q28 Insufficient Balance")

#question29
math, science, english = 75, 68, 82
total = math + science + english
percentage = total / 3
if percentage >= 60:
    print("Q29 Pass")
else:
    print("Q29 Fail")
#question30
purchase = 3500
if purchase > 5000:
    discount = 0.20
elif purchase > 2000:
    discount = 0.10
elif purchase > 1000:
    discount = 0.05
else:
    discount = 0.0
final_amount = purchase * (1 - discount)
print("Q30 Final Amount after Discount:", final_amount)
```
